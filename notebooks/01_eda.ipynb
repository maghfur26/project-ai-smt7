# Exploratory Data Analysis

Notebook ini membaca dataset hasil `scripts/prepare_data.py` dan menampilkan
statistik dasar untuk laporan.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

from src.data.prepare import read_stats

TASKS = ["cyberbullying", "buzzer"]
for task in TASKS:
    path = Path("data/processed") / task / "stats.json"
    if not path.exists():
        print(f"{task}: stats belum ada, jalankan prepare_data.py dulu")
        continue
    print(f"\n=== {task} ===")
    print(read_stats(path.parent))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for axis, task in zip(axes, TASKS):
    directory = Path("data/processed") / task
    if not (directory / "train.csv").exists():
        axis.set_title(f"{task}: data belum siap")
        continue
    train = pd.read_csv(directory / "train.csv")
    train["label"].value_counts().sort_index().plot(kind="bar", ax=axis)
    axis.set_title(f"Distribusi label train — {task}")
    axis.set_xlabel("label")
    axis.set_ylabel("jumlah")
plt.tight_layout()
plt.show()
```

```python
for task in TASKS:
    directory = Path("data/processed") / task
    if not (directory / "train.csv").exists():
        continue
    train = pd.read_csv(directory / "train.csv")
    train["length"] = train["text"].astype(str).str.len()
    print(f"\n=== {task} ===")
    display(train[["source", "label"]].value_counts().head(10))
    display(train["length"].describe())
    display(train[["text", "label"]].sample(5, random_state=42))
```
